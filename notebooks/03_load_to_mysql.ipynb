```python
import pandas as pd
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
```

```python
nutrition = pd.read_csv("../data/cleaned/nutrition_clean.csv")
stores = pd.read_csv("../data/cleaned/stores_clean.csv")

print("nutrition:", nutrition.shape)
print("stores:", stores.shape)
```

```python
load_dotenv()

user = os.getenv("DB_USER")
password = os.getenv("DB_PASSWORD")

engine = create_engine(f"mysql+mysqlconnector://{user}:{password}@localhost/starbucks_analysis")
```

```python
nutrition.to_sql("nutrition", engine, if_exists="append", index=False)
print("nutrition loaded")
```

```python
stores.to_sql("stores", engine, if_exists="append", index=False)
print("stores loaded")
```
