```python
import pandas as pd
df = pd.read_csv('../data/raw/starbucks_nutrition_raw.csv')
df.shape
df.info()
display(df.head())
df.columns.tolist()
df.dtypes
df.isna().sum()

```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "")
    .str.replace(")", "")
    .str.replace("%","pct")
)
df.columns.tolist()
```

```python
#every nutrition column was int however this one was str which pointed to an issue. 
#coerce - it allows to convert all the values to interger and if it is not possible returns na.
#finally mask will store all the locations where the value is na
#.loc takes and gives all the rows with value = True, and the next parameter is the columns
mask = pd.to_numeric(df['total_fat_g'], errors="coerce").isna()
df.loc[mask, "total_fat_g"]
df.loc[237, "total_fat_g"] = "3.2"
df.loc[237, "total_fat_g"]
df["total_fat_g"] = pd.to_numeric(df["total_fat_g"])
df["total_fat_g"].dtype
```

```python
#same way caffeine_mg came to be str and others had 242 non null, this had 241.
df["caffeine_mg"].unique()
df["caffeine_mg"] = pd.to_numeric(df["caffeine_mg"], errors="coerce")
df["caffeine_mg"].unique() 
```

```python
print(df.duplicated().sum())
#no duplicate rows found.
```

```python
pct_cols = ["vitamin_a_pct_dv", "vitamin_c_pct_dv", "calcium_pct_dv", "iron_pct_dv"]

for col in pct_cols:
    df[col] = pd.to_numeric(df[col].str.replace("%", ""), errors="coerce")

df[pct_cols].dtypes
df[pct_cols].isna().sum()
```

```python
df.info()             # every column the right type now?
df.isna().sum()       # where do NaNs remain?
df.shape
```
