```python
import pandas as pd

stores = pd.read_csv("../data/raw/starbucks_stores_raw.csv")

stores.shape
stores.head()
stores.info()
stores.isna().sum()
```

```python
keep_cols = ["StarbucksId", "Name", "OwnershipType",
             "CountrySubdivisionCode", "CountryCode", "Longitude", "Latitude"]
stores = stores[keep_cols]
```

```python
stores.shape        
stores.head()
```

```python
stores = stores.rename(columns={
    "StarbucksId": "store_id",
    "Name": "store_name",
    "OwnershipType": "ownership_type",
    "CountrySubdivisionCode": "state_region",
    "CountryCode": "country_code",
    "Longitude": "longitude",
    "Latitude": "latitude",
})
stores.columns.tolist()
```

```python
# how many distinct country codes, and what are they?
stores["country_code"].nunique()
stores["country_code"].unique()
```

```python
stores.columns.tolist()   # should be 7 columns, no 'city'
stores.shape              # (25600, 7)
```

```python
stores.duplicated().sum()
```

verified no duplicate stores; store_id confirmed unique across all 25,600 rows

```python
stores["store_id"].duplicated().sum()
```

```python
stores[stores["latitude"].isna()]
```

```python
ownership_map = {
    "CO": "Company Owned",
    "LS": "Licensed Store",
    "JV": "Joint Venture",
    "FR": "Franchise"
}
stores["ownership_type"] = stores["ownership_type"].map(ownership_map)
```

```python
stores.info()
print(stores.isna().sum())
print(stores.shape)
```

```python
stores.to_csv("../data/cleaned/stores_clean.csv", index=False)

check_stores = pd.read_csv("../data/cleaned/stores_clean.csv")
check_stores.info()
```
